```python
import numpy as np


def relu(x):
    """ReLU 激活函數"""
    return np.maximum(0, x)

def sigmoid(x):
    """Sigmoid 激活函數"""
    return 1 / (1 + np.exp(-x))

def softmax(x):
    """Softmax 激活函數 (含減去最大值以防數值溢位)"""
    x = x - np.max(x)
    return np.exp(x) / np.sum(np.exp(x))

def cross_entropy_error(y_hat, y):
    """誤差 (Cross Entropy Error)"""
    delta = 1e-7  # 防止 np.log(0) 產生溢位
    return -np.sum(y * np.log(y_hat + delta))





np.random.seed(42)

# --- 輸入層 (Input Layer) ---

x = np.array([2.5, 5, 7.2, 0.8, 6.5, 4.7])
print(f"【輸入層】\n輸入資料 x: {x}\n陣列形狀: {x.shape}\n")

# --- 第一隱藏層 (Hidden Layer 1) ---

W1 = np.random.randn(6, 5) * 0.1
b1 = np.zeros(5)
z1 = np.dot(x, W1) + b1
h1 = relu(z1)

print("【第一隱藏層 (5 個神經元, ReLU)】")
print(f"W1 形狀: {W1.shape}, W1 參數個數: {W1.size}")
print(f"b1 形狀: {b1.shape}, b1 參數個數: {b1.size}")
print(f"加權和 z1: {z1}, 形狀: {z1.shape}")
print(f"輸出 h1: {h1}, 形狀: {h1.shape}\n")

# --- 第二隱藏層 (Hidden Layer 2) ---

W2 = np.random.randn(5, 4) * 0.1
b2 = np.zeros(4)
z2 = np.dot(h1, W2) + b2
h2 = relu(z2)

print("【第二隱藏層 (4 個神經元, ReLU)】")
print(f"W2 形狀: {W2.shape}, W2 參數個數: {W2.size}")
print(f"b2 形狀: {b2.shape}, b2 參數個數: {b2.size}")
print(f"加權和 z2: {z2}, 形狀: {z2.shape}")
print(f"輸出 h2: {h2}, 形狀: {h2.shape}\n")

# --- 第三隱藏層 (Hidden Layer 3) ---

W3 = np.random.randn(4, 4) * 0.1
b3 = np.zeros(4)
z3 = np.dot(h2, W3) + b3
h3 = sigmoid(z3)

print("【第三隱藏層 (4 個神經元, Sigmoid)】")
print(f"W3 形狀: {W3.shape}, W3 參數個數: {W3.size}")
print(f"b3 形狀: {b3.shape}, b3 參數個數: {b3.size}")
print(f"加權和 z3: {z3}, 形狀: {z3.shape}")
print(f"輸出 h3: {h3}, 形狀: {h3.shape}\n")

# --- 輸出層 (Output Layer) ---


W4 = np.random.randn(4, 3) * 0.1
b4 = np.zeros(3)
z4 = np.dot(h3, W4) + b4
y_hat = softmax(z4)

print("【輸出層 (3 個神經元, Softmax)】")
print(f"W4 形狀: {W4.shape}, W4 參數個數: {W4.size}")
print(f"b4 形狀: {b4.shape}, b4 參數個數: {b4.size}")
print(f"加權和 z4: {z4}, 形狀: {z4.shape}")
print(f"輸出 y_hat: {y_hat}, 形狀: {y_hat.shape}\n")

# --- 計算誤差 (Cross Entropy Error) ---
y = np.array([0, 0, 1])
loss = cross_entropy_error(y_hat, y)

print("【誤差計算】")
print(f"真實結果 y: {y}")
print(f"預測結果 y_hat: {y_hat}")
print(f"誤差 (Cross Entropy Error): {loss:.6f}\n")
```

```python
def dnn(x_input):
    x_arr = np.array(x_input)

    # 第一隱藏層 (relu)
    z1_val = np.dot(x_arr, W1) + b1
    h1_val = relu(z1_val)

    # 第二隱藏層 (relu)
    z2_val = np.dot(h1_val, W2) + b2
    h2_val = relu(z2_val)

    # 第三隱藏層 (sigmoid)
    z3_val = np.dot(h2_val, W3) + b3
    h3_val = sigmoid(z3_val)

    # 輸出層 (softmax)
    z4_val = np.dot(h3_val, W4) + b4
    y_hat_val = softmax(z4_val)

    return y_hat_val


y_hat_from_func = dnn(x)
print(f"透過 dnn(x) 計算出的 y_hat: {y_hat_from_func}")
print(f"與第一題結果是否完全相同？ {np.allclose(y_hat, y_hat_from_func)}\n")



def predict_class(f1, f2, f3, f4, f5, f6):

    input_vector = [f1, f2, f3, f4, f5, f6]
    y_hat_res = dnn(input_vector)


    predicted_cls = int(np.argmax(y_hat_res) + 1)

    return f"預測類別：第 {predicted_cls} 類 (最高機率: {y_hat_res[predicted_cls - 1]:.4f})"



inputs = [
    gr.Number(value=1.2,  label="特徵 1 (x1)"),
    gr.Number(value=-0.5, label="特徵 2 (x2)"),
    gr.Number(value=2.1,  label="特徵 3 (x3)"),
    gr.Number(value=0.8,  label="特徵 4 (x4)"),
    gr.Number(value=-1.1, label="特徵 5 (x5)"),
    gr.Number(value=0.5,  label="特徵 6 (x6)")

output = gr.Textbox(label="預測結果")


iface = gr.Interface(
    fn=predict_class,
    inputs=inputs,
    outputs=output,
    title="HW1: 多隱藏層 DNN 分類預測系統",
    description="請在下方輸入框修改或填入 6 個特徵值，點擊 Submit 即可輸出機率最高的類別 (1, 2 或 3)。"
)


iface.launch(share=True)
```
