```python
import numpy as np
```

```python
def relu(x):
    """ReLU 激活函數"""
    return np.maximum(0, x)
```

```python
def sigmoid(x):
    """Sigmoid 激活函數"""
    return 1 / (1 + np.exp(-x))
```

```python
def softmax(x):
    """
    Softmax 函數 (支援批次運算)
    參考「神經網路計算(二)」範例，使用 axis=-1 與 keepdims=True 確保維度正確
    """
    x = x - np.max(x, axis=-1, keepdims=True)
    return np.exp(x) / np.sum(np.exp(x), axis=-1, keepdims=True)
```

```python
def cross_entropy_error(y, t):
    """
    交叉熵誤差 (支援批次運算)
    參考「神經網路計算(二)」範例
    """
    if y.ndim == 1:
        t = t.reshape(1, t.size)
        y = y.reshape(1, y.size)

    batch_size = y.shape[0]
    return -np.sum(t * np.log(y + 1e-7)) / batch_size
```

```python
np.random.seed(42)
```

```python
W1 = np.random.randn(6, 5) * 0.1
b1 = np.zeros(5)
```

```python
W2 = np.random.randn(5, 4) * 0.1
b2 = np.zeros(4)
```

```python
W3 = np.random.randn(4, 4) * 0.1
b3 = np.zeros(4)
```

```python
W4 = np.random.randn(4, 3) * 0.1
b4 = np.zeros(3)
```

```python
X = np.array([
    [2.5, 5.0, 7.2, 0.8, 6.5, 4.7],
    [1.0, 2.3, 3.1, 0.4, 1.5, 2.0],
    [3.2, 1.1, 4.5, 2.2, 3.0, 1.9],
    [0.5, 0.8, 1.2, 0.1, 0.9, 0.4],
    [4.1, 3.5, 6.0, 1.5, 5.2, 3.8]
])
```

```python
Y = np.array([
    [0, 0, 1],
    [1, 0, 0],
    [0, 1, 0],
    [0, 0, 1],
    [1, 0, 0]
])
```

```python
print("【批次輸入與真實資料形狀】")
print(f"輸入矩陣 X 形狀: {X.shape}")
print(f"真實結果 Y 形狀: {Y.shape}\n")
```

```python
print("【各層權重與偏值矩陣形狀】")
print(f"W1 形狀: {W1.shape}, b1 形狀: {b1.shape}")
print(f"W2 形狀: {W2.shape}, b2 形狀: {b2.shape}")
print(f"W3 形狀: {W3.shape}, b3 形狀: {b3.shape}")
print(f"W4 形狀: {W4.shape}, b4 形狀: {b4.shape}\n")
```

```python
Z1 = np.dot(X, W1) + b1
H1 = relu(Z1)
```

```python
Z2 = np.dot(H1, W2) + b2
H2 = relu(Z2)
```

```python
Z3 = np.dot(H2, W3) + b3
H3 = sigmoid(Z3)
```

```python
Z4 = np.dot(H3, W4) + b4
y_hat = softmax(Z4)
```

```python
print("【各層調整後的加權和 z 與輸出向量 h / y_hat 形狀】")
print(f"加權和 Z1 形狀: {Z1.shape} | 輸出 H1 形狀: {H1.shape}")
print(f"加權和 Z2 形狀: {Z2.shape} | 輸出 H2 形狀: {H2.shape}")
print(f"加權和 Z3 形狀: {Z3.shape} | 輸出 H3 形狀: {H3.shape}")
print(f"加權和 Z4 形狀: {Z4.shape} | 輸出 y_hat 形狀: {y_hat.shape}\n")
```

```python
print("預測結果 y_hat:")
print(y_hat)
print()
```

```python
batch_loss = cross_entropy_error(y_hat, Y)

print("【誤差計算】")
print(f"此一整批 (五筆) y_hat 和 Y 誤差值的平均 (Cross Entropy Error): {batch_loss:.6f}")
```
